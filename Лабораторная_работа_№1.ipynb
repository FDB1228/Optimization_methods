**Метод дихитомии**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
def f(x):
  x1= (x ** 2 + x * np.sqrt(3))/(2*x-2)
  x2= (np.sqrt(3)* (x ** 3) - 2 * x + 5)/(7 + np.sqrt(7))
  y=np.arcsin(x1)+x2 ** (np.log(10))
  return y

def dichotomy (a,b,eps):
  d=eps/5
  i=0
  op=0
  n=1
  while b-a>eps:
    plt.plot([a,b],[2*n,2*n],'b',1)
    x1=(a+b)/2 - d
    x2=(a+b)/2 + d
    y1=f(x1)
    y2=f(x2)
    if y1 >= y2:
      b = x2
    else:
      a = x1
    i+=1
    op+=2
    n+=1
  c=(a+b)/2
  print("найденный минимум", c,", значение функции в данной точке ", f(c))
  print("кол-во итераций ",i ,", кол-во вычесленных значений целевой функции",op)
  plt.show()
  return c




tau=(1+np.sqrt(5))/2
x = np.linspace(-1, 0, 100)
y=f(x)
plt.plot(x,y)
plt.show()
d1=dichotomy(-1,0,0.01)
d2=dichotomy(-1,0,1e-8)
d3=dichotomy(-1,0,1e-17)
print('при e:  1e-2               1e-8                  1e-17')
print('___________________________________________________________')
print("max: ", -0.63,"           ",  -0.63463853,"               ","-1.0" )
print("f:     0.52","            ",   0.52793835,"              ", " 0.43243403152459337")
print("i:     ",8, "               ",28, "                    ", "  54" )
print("op:    ",16, "              ",56, "                ","      108")
```

**Метод золотого сечения**

```python
def gold(a,b,eps):
  i=0
  op=2
  n=1
  x1=a+(1-1/tau)*(b-a)
  x2=a+1/tau *(b-a)
  y1=f(x1)
  y2=f(x2)
  while b - a > eps:
    plt.plot([a,b],[2*n,2*n],'b',1)
    n += 1
    if y1 >= y2:
        b = x2
        x2=x1
        y2=y1
        x1=a + b - x2
        y1 = f(x1)
    else:
        a = x1
        y1=y2
        x1=x2
        x2=a+b-x1
        y2=f(x2)
    i += 1
    op+=1
  c = (a+b)/2
  plt.show()
  print("найденный экстремум", c,", значение функции в данной точке ", f(c))
  print("кол-во итераций ",i ,", кол-во вычесленных значений целевой функции",op)
  return c

tau=(1+np.sqrt(5))/2
x = np.linspace(-1, 0, 100)
y=f(x)
plt.plot(x,y)
plt.show()
gold(-1,0,0.01)
gold(-1,0,1e-8)
gold(-1,0,1e-17)

print('при e:  1e-2               1e-8                  1e-17')
print('___________________________________________________________')
print("max:", "-0.63" ,"           ",  -0.63463855,"         ",  -0.6346385634614643 )
print("f:    0.52",    "            ", 0.52793835,"          ",     0.5279383576625023)
print("i:     ",10, "              ",39 , "                  ", 41  )
print("op:    ",12, "              ",56, "                  ",43)
```
