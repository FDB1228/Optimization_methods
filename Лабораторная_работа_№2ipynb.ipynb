```python
# @title Визуализиция графиков квадратичной функции и функции Розенброка
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from typing import Callable, List, Tuple
from numpy.linalg import norm

# Определяем функции
def f1(x: np.ndarray) -> float:
    """Квадаратичная функция"""
    return (
        10 * x[0] ** 2
        - 4 * x[0] * x[1]
        + 7 * x[1] ** 2
        - 4 * np.sqrt(5) * (5 * x[0] - x[1])
        - 16
    )

def f2(x: np.ndarray) -> float:
    """Функция Розенброка alpha = 1"""
    return (x[0] ** 2 - x[1]) ** 2 + (x[0] - 1) ** 2

def f3(x: np.ndarray) -> float:
    """Функция Розенброка alpha = 15"""
    return 15 * (x[0] ** 2 - x[1]) ** 2 + (x[0] - 1) ** 2


# Компоненты градиента
def f1dx(x: np.ndarray) -> float:
    return 20 * x[0] - 4 * x[1] - 20 * np.sqrt(5)

def f1dy(x: np.ndarray) -> float:
    return -4 * x[0] + 14 * x[1] + 4 * np.sqrt(5)

def f2dx(x: np.ndarray) -> float:
    return 4 * x[0] * ((x[0] ** 2) - x[1]) + 2 * (x[0] - 1)

def f2dy(x: np.ndarray) -> float:
    return -2 * ((x[0] ** 2) - x[1])

def f3dx(x: np.ndarray) -> float:
    return 60 * x[0] * ((x[0] ** 2) - x[1]) + 2 * (x[0] - 1)

def f3dy(x: np.ndarray) -> float:
    return -30 * ((x[0] ** 2) - x[1])

# генерация сетки
def generate_mesh_data(f: Callable[[np.ndarray], float],
                       x_range: Tuple[float, float] = (-10, 10),
                       y_range: Tuple[float, float] = (-10, 10),
                       resolution: int = 100) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """Generates mesh grid data for a given function."""
    x = np.linspace(*x_range, resolution)
    y = np.linspace(*y_range, resolution)
    x_grid, y_grid = np.meshgrid(x, y)
    z = np.array([[f(np.array([x, y])) for x in x] for y in y])
    return x_grid, y_grid, z

# трехмерный график функции
def plot_3d_function(f: Callable[[np.ndarray], float], title: str = "3D Function Plot") -> None:
    """Plots a 3D surface of the given function."""
    x_grid, y_grid, z_grid = generate_mesh_data(f)

    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(projection='3d')
    ax.plot_surface(x_grid, y_grid, z_grid, cmap='viridis')

    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("F(X, Y)")
    ax.set_title(title)
    plt.show()


if __name__ == "__main__":
    plot_3d_function(f1, "Function 1: Квадратичная функция")
    plot_3d_function(f2, "Function 2: Функция Розенброка alpha = 1")
    plot_3d_function(f3, "Function 2: Функция Розенброка alpha = 15")

```

```python
# @title Визуализация работы метода наискорейшего спуска
def norm(v: np.ndarray) -> float:
    """Евклидова норма вектора"""
    return np.linalg.norm(v)

def golden_section_search(
    f: Callable[[float], float], a: float, b: float, tol: float) -> float:
    """Метод золотого сечения"""
    phi = (1 + np.sqrt(5)) / 2
    resphi = 2 - phi
    f_evals = 2

    x1, x2 = a + resphi * (b - a), b - resphi * (b - a)
    f1, f2 = f(x1), f(x2)

    while abs(b - a) > tol:
        if f1 < f2:
            b, x2, f2 = x2, x1, f1
            x1 = a + resphi * (b - a)
            f1 = f(x1)
        else:
            a, x1, f1 = x1, x2, f2
            x2 = b - resphi * (b - a)
            f2 = f(x2)
        f_evals+=1
    return (a + b) / 2 , f_evals





def gradient_descent(
    f: Callable[[np.ndarray], float],
    fdx: Callable[[np.ndarray], float],
    fdy: Callable[[np.ndarray], float],
    x0: List[float],
    g: List[float],
    eps: float,
    max_iter: int = 5000,
) -> np.ndarray:

    i = 0
    f_evals = 0
    grad_evals = 1
    opt = np.array(x0, dtype=float)
    w = -np.array([fdx(opt), fdy(opt)])
    norm_w = norm(w)

    # Создание сетки
    fig, ax = plt.subplots(figsize=(8, 8))

    x_vals = np.linspace(g[0][0], g[0][1], 100)
    y_vals = np.linspace(g[1][0], g[1][1], 100)
    X, Y = np.meshgrid(x_vals, y_vals)
    Z = np.array([[f(np.array([x, y])) for x in x_vals] for y in y_vals])

    ax.scatter(*x0, color='blue', label="Начальное приближение")

    path = [opt.copy()]

    draw_iterations = []

    for iteration in range(max_iter):
        if norm_w <= eps:
            break


        phi = lambda xi: f(opt + xi * w)

        # определение оптимального параметра xi
        xi, f_evals_k = golden_section_search(phi, 0, 0.7, 1e-10)

        # обновление переменных
        x_start, y_start = opt[0], opt[1]
        opt += xi * w
        path.append(opt.copy())
        w = -np.array([fdx(opt), fdy(opt)])
        norm_w = norm(w)
        x_end, y_end = opt[0], opt[1]
        i += 1
        f_evals += f_evals_k
        grad_evals += 1

        # Отрисовка линий уровня
        if i <= 10:
            draw_iterations.append(i)
        else:
            if i % 30 == 0 or iteration == max_iter - 1 or norm_w <= eps:
                draw_iterations.append(i)


        ax.quiver(x_start, y_start, x_end - x_start, y_end - y_start,
                 angles='xy', scale_units='xy', scale=1, color='black', alpha=1, width = 0.005, headwidth = 2, zorder = 3)


        if i in draw_iterations:
            current_level = f(opt)
            cs = ax.contour(X, Y, Z, levels=[current_level], colors='green', linewidths=1)


    # Визуализация
    ax.scatter(*opt, color='red', label="Точка минимума", zorder = 3)


    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.legend()
    ax.set_title("Метод наискорейшего спуска ")
    plt.grid()
    plt.show()

    print(f"Начальное приближение: ({x0[0]},{x0[1]})")
    if eps == 1e-2:
        print(f"Точка минимума: x = {opt[0]:.2f}, y = {opt[1]:.2f}")
        print(f"Значение функции f(x,y)={f(opt):.2f}")
    else:
        print(f"Точка минимума: x = {opt[0]:.5f}, y = {opt[1]:.5f}")
        print(f"Значение функции f(x,y)={f(opt):.5f}")

    print(f"Количество вычислений функции: {f_evals}")
    print(f"Количество вычислений градиента: {grad_evals}")
    print(f"Количество итераций: {i}")

    info = [x0, eps, i, f_evals, opt, f(opt)]

    return info



if __name__ == "__main__":
    x1 = [5, -2]
    x2 = [-5, 10]

    eps1 = 1e-2
    eps2 = 1e-5

    g11 = [[2, 5.5],[-2.2, 0.5]]
    g12 = [[-5.2, 5],[0.5, 10.5]]

    g21 = [[0, 5.2],[-2.5, 1.5]]
    g22 = [[-5.2, 3],[0.5, 11]]

    g31 = [[-0.2, 5.2],[-2.2, 1.2]]
    g32 = [[-5.2, 4],[0, 11]]


    inf1 = gradient_descent(f1, f1dx, f1dy, x1, g11, eps1)
    inf2 = gradient_descent(f1, f1dx, f1dy, x1, g11, eps2)

    inf3 = gradient_descent(f1, f1dx, f1dy, x2, g12, eps1)
    inf4 = gradient_descent(f1, f1dx, f1dy, x2, g12, eps2)

    inf5 = gradient_descent(f2, f2dx, f2dy, x1, g21 ,eps1)
    inf6 = gradient_descent(f2, f2dx, f2dy, x1, g21, eps2)

    inf7 = gradient_descent(f2, f2dx, f2dy, x2, g22, eps1)
    inf8 = gradient_descent(f2, f2dx, f2dy, x2, g22, eps2)

    inf9 = gradient_descent(f3, f3dx, f3dy, x1, g31 ,eps1)
    inf10 = gradient_descent(f3, f3dx, f3dy, x1, g31, eps2)

    inf11 = gradient_descent(f3, f3dx, f3dy, x2, g32, eps1)
    inf12 = gradient_descent(f3, f3dx, f3dy, x2, g32, eps2)









```

```python
# @title Визуализиция работы метода дробления шага
def gradient_step_descent(
    f: Callable[[np.ndarray], float],
    fdx: Callable[[np.ndarray], float],
    fdy: Callable[[np.ndarray], float],
    x0: List[float],
    g: List[float],
    eps: float,
    max_iter: int = 40000,
    gamma: float = 0.5,
    omega: float = 0.1,
    xi: float = 1.0
) -> np.ndarray:

    i = 0
    f_evals = 0
    grad_evals = 0
    opt = np.array(x0, dtype=float)
    grad = np.array([fdx(opt), fdy(opt)])
    norm_grad = norm(grad)
    grad_evals += 1

    # Создание сетки
    fig, ax = plt.subplots(figsize=(8, 8))

    x_vals = np.linspace(g[0][0], g[0][1], 100)
    y_vals = np.linspace(g[1][0], g[1][1], 100)
    X, Y = np.meshgrid(x_vals, y_vals)
    Z = np.array([[f(np.array([x, y])) for x in x_vals] for y in y_vals])

    ax.scatter(*x0, color='blue', label="Начальное приближение")

    path = [opt.copy()]

    for _ in range(max_iter):
        if norm_grad <= eps:
            break

        f_evals += 1

        # Дробление шага
        while True:
            x_new = opt - xi * grad
            f_new = f(x_new)
            f_evals += 1
            if f_new <= f(opt) - omega * xi * norm_grad**2:
                break
            xi *= gamma
            if xi < 1e-10:
                break

        # Обновление переменных
        x_start, y_start = opt[0], opt[1]
        opt = x_new
        path.append(opt.copy())
        grad = np.array([fdx(opt), fdy(opt)])
        norm_grad = norm(grad)
        grad_evals += 1
        x_end, y_end = opt[0], opt[1]
        i += 1

        # Визуализация на каждой итерации
        ax.quiver(
            x_start, y_start, x_end - x_start, y_end - y_start,
            angles='xy', scale_units='xy', scale=1, color='black', alpha=1,
            width=0.005, headwidth=2, zorder=3
        )

        # Отрисовка линий уровня
        if i <= 30 or i % 30 == 0 or _ == max_iter - 1 or norm_grad <= eps:
            current_level = f(opt)
            cs = ax.contour(X, Y, Z, levels=[current_level], colors='green', linewidths=1)

    # Визуализация
    ax.scatter(*opt, color='red', s=50, linewidths=2, label="Точка минимума", zorder=10)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.legend()
    ax.set_title(f"Метод дробления шага")
    plt.grid()
    plt.show()

    # Вывод результатов
    print(f"Начальное приближение: ({x0[0]},{x0[1]})")
    if eps == 1e-2:
        print(f"Точка минимума: x = {opt[0]:.2f}, y = {opt[1]:.2f}")
    else:
        print(f"Точка минимума: x = {opt[0]:.5f}, y = {opt[1]:.5f}")
    print(f"Количество вычислений функции: {f_evals}")
    print(f"Количество вычислений градиента: {grad_evals}")
    print(f"Количество итераций: {i}")

    info = [x0, eps, i, f_evals, opt, f(opt)]

    return info

if __name__ == "__main__":
    x1 = [5, -2]
    x2 = [-5, 10]

    eps1 = 1e-2
    eps2 = 1e-5

    g11 = [[0.5, 5.2],[-2.2, 0.5]]
    g12 = [[-5.2, 6.5],[-1, 10.2]]

    g21 = [[-4, 5.2],[-2.2, 1.5]]
    g22 = [[-5.2, 5.5],[0.5, 12]]

    g31 = [[-3.2, 5.2],[-2.2, 2]]
    g32 = [[-5.2, 4],[0, 11]]


    i1 = gradient_step_descent(f1, f1dx, f1dy, x1, g11, eps1)
    i2 = gradient_step_descent(f1, f1dx, f1dy, x1, g11, eps2)

    i3 = gradient_step_descent(f1, f1dx, f1dy, x2, g12, eps1)
    i4 = gradient_step_descent(f1, f1dx, f1dy, x2, g12, eps2)

    i5 = gradient_step_descent(f2, f2dx, f2dy, x1, g21, eps1)
    i6 = gradient_step_descent(f2, f2dx, f2dy, x1, g21, eps2)

    i7 = gradient_step_descent(f2, f2dx, f2dy, x2, g22, eps1)
    i8 = gradient_step_descent(f2, f2dx, f2dy, x2, g22, eps2)

    i9 = gradient_step_descent(f3, f3dx, f3dy, x1, g31,eps1)
    i10 = gradient_step_descent(f3, f3dx, f3dy, x1, g31, eps2)

    i11 = gradient_step_descent(f3, f3dx, f3dy, x2, g32, eps1)
    i12 = gradient_step_descent(f3, f3dx, f3dy, x2, g32, eps2)
```

Объясняется это тем, что при использовании МЗС минимизируется ф-ия $f(X^{K-1} + \chi \omega^k)$. Последняя будет унимодальна в случае выпуклой функции и не унимодальна в случаи не выпуклой функции, т.е. МЗС будет находить 2 различные точки минимума. Таким образом, выпуклость функции влияет на стоимость методов.
3. Влияние овражности. Сравним стоимость методов на функции Розенброка при

Для функции Розенброка при $\alpha = 15$ метод наискорейшего спуска требует в среднем в $6.5$ раз больше вычислений функции чем метод дробления шага. Таким образом, метод дробления шага дешевле чем метод наискорейшего спуска как для выпуклых функций, так и для невыпуклых.
